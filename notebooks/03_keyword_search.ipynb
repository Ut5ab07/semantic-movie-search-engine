# Baseline Keyword Search (TF-IDF)

This notebook implements a traditional keyword-based retrieval system
using TF-IDF vectorization and cosine similarity.

This will later be compared with semantic embedding search.


## Importing Libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## Loading Processed Data

```python
df = pd.read_csv("../data/processed/movies_semantic.csv")
df.head()
```

```python
df.shape
```

## Creating TF-IDF Vectorizer

```python
vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=10000
)
```

```python
tfidf_matrix = vectorizer.fit_transform(df['semantic_text'])
```

```python
tfidf_matrix.shape
```

## Building Search Function

```python
def keyword_search(query, top_n=5):
    query_vec = vectorizer.transform([query])
    
    similarities = cosine_similarity(query_vec, tfidf_matrix)
    
    top_indices = similarities[0].argsort()[-top_n:][::-1]
    
    results = df.iloc[top_indices][['title', 'vote_average']]
    
    return results
```

## Testing Function

```python
keyword_search("unique alien mission")
```

```python
keyword_search("italian mafia family business")
```

```python
keyword_search("loneliness in space")
```
